```python
import pandas as pd
import numpy as np
import math
from IPython.display import display
import ace_tools_open as tools







# Read the CSV file (using ";" as the delimiter)
df  = pd.read_csv("OneDrive/ACIT/Master Thesis/JupytherLab/cloudsim-custom-implementation/logs/PA_Storage.csv", delimiter=";")

# Display the table
df .head()  # Shows the first 5 rows
```

```python
# === STEP 1: Convert data types ===
df['cpu_power'] = df['cpu_power'].astype(float)
df['ram_power'] = df['ram_power'].astype(float)
df['bw_power'] = df['bw_power'].astype(float)
df['I/O_power'] = df['I/O_power'].astype(float)

df['mips'] = df['mips'].astype(float)
df['available_mips'] = df['available_mips'].astype(float)
df['cpu_utilization'] = df['cpu_utilization'].astype(float)

df['ram'] = df['ram'].astype(float)
df['available_ram'] = df['available_ram'].astype(float)
df['bw'] = df['bw'].astype(float)
df['available_bw'] = df['available_bw'].astype(float)
df['storage'] = df['storage'].astype(float)
df['available_storage'] = df['available_storage'].astype(float)
df['disk_I/O'] = df['disk_I/O'].astype(float)

# === STEP 2: Collect columns into lists ===
columns_as_lists = {col: df[col].tolist() for col in df.columns}

# === STEP 3: Filter active and powered-on hosts ===
active_hosts_df = df[df['active'] == True]
power_on_hosts_df = df[df['power_on'] == True]

# === STEP 4: Usage percentages ===
def usage_percentage(available, total):
    return [(t - a) / t * 100 if t else 0 for a, t in zip(available, total)]

cpu_usage_pct = usage_percentage(df['available_mips'], df['mips'])
ram_usage_pct = usage_percentage(df['available_ram'], df['ram'])
bw_usage_pct = usage_percentage(df['available_bw'], df['bw'])
storage_usage_pct = usage_percentage(df['available_storage'], df['storage'])

# === STEP 5: Totals and per-host aggregation ===
total_cpu_power = df['cpu_power'].sum()
total_ram_power = df['ram_power'].sum()
total_bw_power = df['bw_power'].sum()
total_io_power = df['I/O_power'].sum()
total_disk_io = df['disk_I/O'].sum()

# === STEP 6: VM Parsing ===
def parse_vms(vms_str):
    vms_data = []
    if pd.isna(vms_str) or vms_str.strip() == '':
        return vms_data
    for vm_entry in vms_str.strip(':').split(':'):
        try:
            dc_id, mips, ram, bw = vm_entry.split(',')
            vms_data.append({
                'datacenter_id': int(dc_id),
                'allocated_mips': float(mips),
                'requested_ram': float(ram),
                'requested_bw': float(bw)
            })
        except:
            continue
    return vms_data

all_vms = []
for index, row in df.iterrows():
    vms = parse_vms(row['vms'])
    for vm in vms:
        vm['time'] = row['time']
        vm['host_id'] = row['host_id']
        all_vms.append(vm)

vms_df = pd.DataFrame(all_vms)
vm_summary_per_host = vms_df.groupby('host_id')[['allocated_mips', 'requested_ram', 'requested_bw']].sum().reset_index()

# === STEP 7: Datacenter-wide Totals and Usage ===
total_mips = df['mips'].sum()
available_mips = df['available_mips'].sum()
mips_usage_pct_dc = (total_mips - available_mips) / total_mips * 100 if total_mips else 0

total_ram = df['ram'].sum()
available_ram = df['available_ram'].sum()
ram_usage_pct_dc = (total_ram - available_ram) / total_ram * 100 if total_ram else 0

total_bw = df['bw'].sum()
available_bw = df['available_bw'].sum()
bw_usage_pct_dc = (total_bw - available_bw) / total_bw * 100 if total_bw else 0

total_storage = df['storage'].sum()
available_storage = df['available_storage'].sum()
storage_usage_pct_dc = (total_storage - available_storage) / total_storage * 100 if total_storage else 0

# === STEP 8: Interpolated Energy ===
df['time_hours'] = df['time'] / 3600
total_cpu_energy_kwh = 0.0
total_ram_energy_kwh = 0.0
total_bw_energy_kwh = 0.0
total_io_energy_kwh = 0.0

for host_id, group in df.groupby('host_id'):
    group_sorted = group.sort_values(by='time_hours')
    times = group_sorted['time_hours'].values
    cpu_powers = group_sorted['cpu_power'].values
    ram_powers = group_sorted['ram_power'].values
    bw_powers = group_sorted['bw_power'].values
    io_powers = group_sorted['I/O_power'].values

    for i in range(1, len(times)):
        delta_t = times[i] - times[i - 1]
        total_cpu_energy_kwh += 0.5 * (cpu_powers[i] + cpu_powers[i - 1]) * delta_t / 1000
        total_ram_energy_kwh += 0.5 * (ram_powers[i] + ram_powers[i - 1]) * delta_t / 1000
        total_bw_energy_kwh += 0.5 * (bw_powers[i] + bw_powers[i - 1]) * delta_t / 1000
        total_io_energy_kwh += 0.5 * (io_powers[i] + io_powers[i - 1]) * delta_t / 1000

# === STEP 9: Summary Output ===
summary = {
    "Total Active Hosts (Running)": active_hosts_df['host_id'].nunique(),
    "Total Hosts Powered On": power_on_hosts_df['host_id'].nunique(),
    "Total VMs Parsed": len(vms_df),
    "Total CPU Power (W)": total_cpu_power,
    "Total RAM Power (W)": total_ram_power,
    "Total BW Power (W)": total_bw_power,
    "Total I/O Power (W)": total_io_power,
    "Total Disk I/O Throughput": total_disk_io,
    "Total CPU Energy (kWh)": total_cpu_energy_kwh,
    "Total RAM Energy (kWh)": total_ram_energy_kwh,
    "Total BW Energy (kWh)": total_bw_energy_kwh,
    "Total I/O Energy (kWh)": total_io_energy_kwh,
    "Datacenter MIPS Usage (%)": mips_usage_pct_dc,
    "Datacenter RAM Usage (%)": ram_usage_pct_dc,
    "Datacenter BW Usage (%)": bw_usage_pct_dc,
    "Datacenter Storage Usage (%)": storage_usage_pct_dc,
}

summary_df = pd.DataFrame([summary])
summary_df = pd.DataFrame([summary]).T.reset_index()
summary_df.columns = ['Metric', 'Value']
#print(summary_df.to_string(index=False))
tools.display_dataframe_to_user(name="Final Simulation Summary Table", dataframe=summary_df)
```

```python
# Make sure time is in hours
df['time_hours'] = df['time'] / 3600

idle_energy_kwh = 0.0
active_energy_kwh = 0.0

# Group by host and integrate each separately
for host_id, group in df.groupby('host_id'):
    group = group.sort_values(by='time_hours')
    times = group['time_hours'].values
    powers = group['cpu_power'].values
    activity = group['active'].values  # True/False

    for i in range(1, len(times)):
        delta_t = times[i] - times[i - 1]
        avg_power = 0.5 * (powers[i] + powers[i - 1])

        # Check if both samples are active or idle
        if activity[i] and activity[i - 1]:
            active_energy_kwh += (avg_power * delta_t) / 1000
        elif not activity[i] and not activity[i - 1]:
            idle_energy_kwh += (avg_power * delta_t) / 1000
        else:
            # One is active, the other idle — split contribution
            active_energy_kwh += (avg_power * delta_t * 0.5) / 1000
            idle_energy_kwh += (avg_power * delta_t * 0.5) / 1000

# Output results
print(f"Interpolated ACTIVE CPU energy: {active_energy_kwh:.6f} kWh")
print(f"Interpolated IDLE  CPU energy: {idle_energy_kwh:.6f} kWh")
```

```python
# --- Step 1: Extract idle and max power per host from power_model column
def extract_idle_max(power_model_str):
    if pd.isna(power_model_str):
        return (np.nan, np.nan)
    try:
        power_vals = [float(x) for x in power_model_str.strip(',').split(',')]
        return (power_vals[0], power_vals[-1])  # idle, max
    except:
        return (np.nan, np.nan)

df[['p_idle', 'p_max']] = df['power_model'].apply(lambda s: pd.Series(extract_idle_max(s)))

# --- Step 2: Calculate MIPS-based utilization
df['mips_utilization'] = (df['mips'] - df['available_mips']) / df['mips']

# --- Step 3: Estimate CPU power from utilization and power model
df['estimated_cpu_power'] = df['p_idle'] + (df['p_max'] - df['p_idle']) * df['mips_utilization']

# --- Step 4: Convert time to hours
df['time_hours'] = df['time'] / 3600

# --- Step 5: Interpolate energy per host, grouped by datacenter
datacenter_energy = {}

for (dc_id, host_id), group in df.groupby(['datacenter_id', 'host_id']):
    group_sorted = group.sort_values(by='time_hours')
    times = group_sorted['time_hours'].values
    powers = group_sorted['estimated_cpu_power'].values

    host_energy_kwh = 0.0
    for i in range(1, len(times)):
        delta_t = times[i] - times[i - 1]
        avg_power = 0.5 * (powers[i] + powers[i - 1])
        host_energy_kwh += (avg_power * delta_t) / 1000

    # Add to datacenter total
    if dc_id not in datacenter_energy:
        datacenter_energy[dc_id] = 0.0
    datacenter_energy[dc_id] += host_energy_kwh

# --- Step 6: Print per-datacenter totals and overall total
print("=== Estimated CPU Energy by Datacenter (kWh) ===")
total_energy_kwh = 0.0
for dc_id, energy in datacenter_energy.items():
    print(f"Datacenter {dc_id}: {energy:.6f} kWh")
    total_energy_kwh += energy

print(f"\nTotal Estimated CPU Energy (All Datacenters): {total_energy_kwh:.6f} kWh")
```
