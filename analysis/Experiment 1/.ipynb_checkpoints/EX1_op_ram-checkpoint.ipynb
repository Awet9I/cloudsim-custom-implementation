```python
import pandas as pd
import numpy as np
import math
from IPython.display import display
import ace_tools_open as tools






# Read the CSV file (using ";" as the delimiter)
df  = pd.read_csv("OneDrive/ACIT/Master Thesis/JupytherLab/cloudsim-custom-implementation/logs/cpu_dvfs_ram.csv", delimiter=";")

# Display the table
df .head()  # Shows the first 5 rows
```

```python
# === STEP 1: Convert data types ===
df['cpu_power'] = df['cpu_power'].astype(float)
df['ram_power'] = df['ram_power'].astype(float)
df['bw_power'] = df['bw_power'].astype(float)
df['I/O_power'] = df['I/O_power'].astype(float)

df['mips'] = df['mips'].astype(float)
df['available_mips'] = df['available_mips'].astype(float)
df['cpu_utilization'] = df['cpu_utilization'].astype(float)

df['ram'] = df['ram'].astype(float)
df['available_ram'] = df['available_ram'].astype(float)
df['bw'] = df['bw'].astype(float)
df['available_bw'] = df['available_bw'].astype(float)
df['storage'] = df['storage'].astype(float)
df['available_storage'] = df['available_storage'].astype(float)
df['disk_I/O'] = df['disk_I/O'].astype(float)

# === STEP 2: Loop through each datacenter ===
summaries = []
for dc_name, group in df.groupby('datacenter_name'):
    active_hosts_df = group[group['active'] == True]
    power_on_hosts_df = group[group['power_on'] == True]

    def usage_percentage(available, total):
        return [(t - a) / t * 100 if t else 0 for a, t in zip(available, total)]

    cpu_usage_pct = usage_percentage(group['available_mips'], group['mips'])
    ram_usage_pct = usage_percentage(group['available_ram'], group['ram'])
    bw_usage_pct = usage_percentage(group['available_bw'], group['bw'])
    storage_usage_pct = usage_percentage(group['available_storage'], group['storage'])

    total_cpu_power = group['cpu_power'].sum()
    total_ram_power = group['ram_power'].sum()
    total_bw_power = group['bw_power'].sum()
    total_io_power = group['I/O_power'].sum()
    total_disk_io = group['disk_I/O'].sum()

    def parse_vms(vms_str):
        vms_data = []
        if pd.isna(vms_str) or vms_str.strip() == '':
            return vms_data
        for vm_entry in vms_str.strip(':').split(':'):
            try:
                dc_id, mips, ram, bw = vm_entry.split(',')
                vms_data.append({
                    'datacenter_id': int(dc_id),
                    'allocated_mips': float(mips),
                    'requested_ram': float(ram),
                    'requested_bw': float(bw)
                })
            except:
                continue
        return vms_data

    all_vms = []
    for index, row in group.iterrows():
        vms = parse_vms(row['vms'])
        for vm in vms:
            vm['time'] = row['time']
            vm['host_id'] = row['host_id']
            all_vms.append(vm)

    vms_df = pd.DataFrame(all_vms)

    total_mips = group['mips'].sum()
    available_mips = group['available_mips'].sum()
    mips_usage_pct_dc = (total_mips - available_mips) / total_mips * 100 if total_mips else 0

    total_ram = group['ram'].sum()
    available_ram = group['available_ram'].sum()
    ram_usage_pct_dc = (total_ram - available_ram) / total_ram * 100 if total_ram else 0

    total_bw = group['bw'].sum()
    available_bw = group['available_bw'].sum()
    bw_usage_pct_dc = (total_bw - available_bw) / total_bw * 100 if total_bw else 0

    total_storage = group['storage'].sum()
    available_storage = group['available_storage'].sum()
    storage_usage_pct_dc = (total_storage - available_storage) / total_storage * 100 if total_storage else 0

    group['time_hours'] = group['time'] / 3600
    total_cpu_energy_kwh = 0.0
    total_ram_energy_kwh = 0.0
    total_bw_energy_kwh = 0.0
    total_io_energy_kwh = 0.0

    for host_id, subgroup in group.groupby('host_id'):
        subgroup_sorted = subgroup.sort_values(by='time_hours')
        times = subgroup_sorted['time_hours'].values
        cpu_powers = subgroup_sorted['cpu_power'].values
        ram_powers = subgroup_sorted['ram_power'].values
        bw_powers = subgroup_sorted['bw_power'].values
        io_powers = subgroup_sorted['I/O_power'].values

        for i in range(1, len(times)):
            delta_t = times[i] - times[i - 1]
            total_cpu_energy_kwh += 0.5 * (cpu_powers[i] + cpu_powers[i - 1]) * delta_t / 1000
            total_ram_energy_kwh += 0.5 * (ram_powers[i] + ram_powers[i - 1]) * delta_t / 1000
            total_bw_energy_kwh += 0.5 * (bw_powers[i] + bw_powers[i - 1]) * delta_t / 1000
            total_io_energy_kwh += 0.5 * (io_powers[i] + io_powers[i - 1]) * delta_t / 1000

    used_mips = total_mips - available_mips
    cpu_utilization_pct = (used_mips / total_mips * 100) if total_mips else 0
    avg_cpu_util_column = group['cpu_utilization'].mean() * 100

    summary = {
        "Datacenter": dc_name,
        "Total Active Hosts (Running)": active_hosts_df['host_id'].nunique(),
        "Total Hosts Powered On": power_on_hosts_df['host_id'].nunique(),
        "Total VMs Parsed": len(vms_df),
        "Total CPU Power (W)": total_cpu_power,
        "Total RAM Power (W)": total_ram_power,
        "Total BW Power (W)": total_bw_power,
        "Total I/O Power (W)": total_io_power,
        "Total Disk I/O Throughput": total_disk_io,
        "Total CPU Energy (kWh)": total_cpu_energy_kwh,
        "Total RAM Energy (kWh)": total_ram_energy_kwh,
        "Total BW Energy (kWh)": total_bw_energy_kwh,
        "Total I/O Energy (kWh)": total_io_energy_kwh,
        "Datacenter MIPS Usage (%)": mips_usage_pct_dc,
        "Datacenter RAM Usage (%)": ram_usage_pct_dc,
        "Datacenter BW Usage (%)": bw_usage_pct_dc,
        "Datacenter Storage Usage (%)": storage_usage_pct_dc,
        "CPU Util from MIPS (%)": cpu_utilization_pct,
        "Average CPU Util from Column (%)": avg_cpu_util_column
    }

    summaries.append(summary)

summary_df = pd.DataFrame(summaries)
summary_pivot = summary_df.set_index('Datacenter').T.reset_index()
summary_pivot.columns.values[0] = 'Metric'
tools.display_dataframe_to_user(name="DVFS + RAM DVFS Summary", dataframe=summary_pivot)
```

```python
# === STEP 1: Convert data types early ===
df['cpu_power'] = df['cpu_power'].astype(float)
df['mips'] = df['mips'].astype(float)
df['available_mips'] = df['available_mips'].astype(float)
df['ram'] = df['ram'].astype(float)
df['available_ram'] = df['available_ram'].astype(float)
df['bw'] = df['bw'].astype(float)
df['available_bw'] = df['available_bw'].astype(float)
df['storage'] = df['storage'].astype(float)
df['available_storage'] = df['available_storage'].astype(float)

# === STEP 2: Collect all columns into individual lists ===
columns_as_lists = {col: df[col].tolist() for col in df.columns}

# === STEP 3: Filter active hosts ===
active_hosts_df = df[df['active'] == True]

# === STEP 4: Usage Percentage Calculations ===
def usage_percentage(available, total):
    return [(t - a) / t * 100 if t else 0 for a, t in zip(available, total)]

cpu_usage_pct = usage_percentage(df['available_mips'], df['mips'])
ram_usage_pct = usage_percentage(df['available_ram'], df['ram'])
bw_usage_pct = usage_percentage(df['available_bw'], df['bw'])
storage_usage_pct = usage_percentage(df['available_storage'], df['storage'])

# === STEP 5: CPU Power Analysis ===
total_cpu_power = df['cpu_power'].sum()
cpu_power_per_host = df.groupby('host_id')['cpu_power'].sum().to_dict()

# === STEP 6: VM Parsing and Analysis ===
def parse_vms(vms_str):
    vms_data = []
    if pd.isna(vms_str) or vms_str.strip() == '':
        return vms_data
    for vm_entry in vms_str.strip(':').split(':'):
        try:
            dc_id, mips, ram, bw = vm_entry.split(',')
            vms_data.append({
                'datacenter_id': int(dc_id),
                'allocated_mips': float(mips),
                'requested_ram': float(ram),
                'requested_bw': float(bw)
            })
        except:
            continue
    return vms_data

all_vms = []
for index, row in df.iterrows():
    vms = parse_vms(row['vms'])
    for vm in vms:
        vm['time'] = row['time']
        vm['host_id'] = row['host_id']
        all_vms.append(vm)



# === CPU POWER CONSUMPTION USING INTERPOLATION ===

# Ensure sorted by time
df_sorted = df.sort_values(by='time')

# Convert time from seconds to hours
time_hours = df_sorted['time'] / 3600
power_watts = df_sorted['cpu_power']

# Trapezoidal integration: compute energy between each pair of points
cpu_energy_kwh = 0.0
for i in range(1, len(df_sorted)):
    t1 = time_hours.iloc[i - 1]
    t2 = time_hours.iloc[i]
    p1 = power_watts.iloc[i - 1]
    p2 = power_watts.iloc[i]
    
    delta_t = t2 - t1  # time in hours
    area = 0.5 * (p1 + p2) * delta_t  # area under the curve segment (Wh)
    cpu_energy_kwh += area / 1000  # convert Wh to kWh


# === INTERPOLATED TOTAL ENERGY ACROSS ALL HOSTS ===

total_energy_kwh = 0.0

# Make sure time is in hours
df['time_hours'] = df['time'] / 3600

# Group by host and compute per-host energy using trapezoidal rule
for host_id, group in df.groupby('host_id'):
    group_sorted = group.sort_values(by='time_hours')
    times = group_sorted['time_hours'].values
    powers = group_sorted['cpu_power'].values

    # Interpolated trapezoidal integration
    host_energy_wh = 0.0
    for i in range(1, len(times)):
        delta_t = times[i] - times[i - 1]
        avg_power = 0.5 * (powers[i] + powers[i - 1])
        host_energy_wh += avg_power * delta_t  # in Wh

    total_energy_kwh += host_energy_wh / 1000  # Wh to kWh






vms_df = pd.DataFrame(all_vms)
vm_summary_per_host = vms_df.groupby('host_id')[['allocated_mips', 'requested_ram', 'requested_bw']].sum().reset_index()

# === STEP 7: Datacenter-wide Totals & Usage ===
total_mips = df['mips'].sum()
available_mips = df['available_mips'].sum()
mips_usage_pct_dc = (total_mips - available_mips) / total_mips * 100 if total_mips else 0

total_ram = df['ram'].sum()
available_ram = df['available_ram'].sum()
ram_usage_pct_dc = (total_ram - available_ram) / total_ram * 100 if total_ram else 0

total_bw = df['bw'].sum()
available_bw = df['available_bw'].sum()
bw_usage_pct_dc = (total_bw - available_bw) / total_bw * 100 if total_bw else 0

total_storage = df['storage'].sum()
available_storage = df['available_storage'].sum()
storage_usage_pct_dc = (total_storage - available_storage) / total_storage * 100 if total_storage else 0


# 1. Count of active hosts
num_active_hosts = active_hosts_df['host_id'].nunique()

# 2. Total number of VMs parsed
total_vms = len(vms_df)

# 3. Number of VMs per host
vms_per_host = vms_df.groupby('host_id').size().reset_index(name='vm_count')


# === DISPLAY ANALYSIS RESULTS ===
print("=== Host-Level Resource Usage Summary ===")
print(f"Total active hosts: {len(active_hosts_df)}")
print(f"Average CPU usage per host (%): {sum(cpu_usage_pct) / len(cpu_usage_pct):.2f}")
print(f"Average RAM usage per host (%): {sum(ram_usage_pct) / len(ram_usage_pct):.2f}")
print(f"Average BW usage per host (%): {sum(bw_usage_pct) / len(bw_usage_pct):.2f}")
print(f"Average Storage usage per host (%): {sum(storage_usage_pct) / len(storage_usage_pct):.2f}")

print("\n=== CPU Power Analysis ===")
print(f"Total CPU Power: {total_cpu_power:.2f}")
print("CPU Power Per Host:")
#for host, power in cpu_power_per_host.items():
    #print(f"  Host {host}: {power:.2f}")

print("\n=== VM Resource Usage Summary Per Host ===")
print(vm_summary_per_host)

print("\n=== Datacenter-Wide Totals and Usage ===")
print(f"Total MIPS: {total_mips:.2f}, Usage: {mips_usage_pct_dc:.2f}%")
print(f"Total RAM: {total_ram:.2f}, Usage: {ram_usage_pct_dc:.2f}%")
print(f"Total Bandwidth: {total_bw:.2f}, Usage: {bw_usage_pct_dc:.2f}%")
print(f"Total Storage: {total_storage:.2f}, Usage: {storage_usage_pct_dc:.2f}%")

# === PRINT RESULTS ===
print("\n=== VM and Host Summary ===")
print(f"Number of active hosts: {num_active_hosts}")
print(f"Total VMs parsed: {total_vms}")
print("Number of VMs hosted by each host:")
print(vms_per_host)





# === PRINT ENERGY ===
print("\n=== CPU Power Consumption (Interpolated) ===")
print(f"Simulation duration: {df_sorted['time'].min():.2f} sec → {df_sorted['time'].max():.2f} sec")
print(f"Interpolated CPU energy usage: {cpu_energy_kwh:.6f} kWh")



# === PRINT TOTAL ENERGY ===
print("\n=== TOTAL Interpolated CPU Energy for All Hosts ===")
print(f"Estimated total CPU energy usage: {total_energy_kwh:.6f} kWh")
```

```python
# === STEP 1: Convert data types early ===
df['cpu_power'] = df['cpu_power'].astype(float)
df['ram_power'] = df['ram_power'].astype(float)  # New field
df['mips'] = df['mips'].astype(float)
df['available_mips'] = df['available_mips'].astype(float)
df['ram'] = df['ram'].astype(float)
df['available_ram'] = df['available_ram'].astype(float)
df['bw'] = df['bw'].astype(float)
df['available_bw'] = df['available_bw'].astype(float)
df['storage'] = df['storage'].astype(float)
df['available_storage'] = df['available_storage'].astype(float)

# === STEP 2: Collect all columns into individual lists ===
columns_as_lists = {col: df[col].tolist() for col in df.columns}

# === STEP 3: Filter active hosts (running hosts)
active_hosts_df = df[df['active'] == True]

# === STEP 4: Usage Percentage Calculations ===
def usage_percentage(available, total):
    return [(t - a) / t * 100 if t else 0 for a, t in zip(available, total)]

cpu_usage_pct = usage_percentage(df['available_mips'], df['mips'])
ram_usage_pct = usage_percentage(df['available_ram'], df['ram'])
bw_usage_pct = usage_percentage(df['available_bw'], df['bw'])
storage_usage_pct = usage_percentage(df['available_storage'], df['storage'])

# === STEP 5: CPU and RAM Power Analysis ===
total_cpu_power = df['cpu_power'].sum()
cpu_power_per_host = df.groupby('host_id')['cpu_power'].sum().to_dict()

total_ram_power = df['ram_power'].sum()
ram_power_per_host = df.groupby('host_id')['ram_power'].sum().to_dict()

# === STEP 6: VM Parsing and Analysis ===
def parse_vms(vms_str):
    vms_data = []
    if pd.isna(vms_str) or vms_str.strip() == '':
        return vms_data
    for vm_entry in vms_str.strip(':').split(':'):
        try:
            dc_id, mips, ram, bw = vm_entry.split(',')
            vms_data.append({
                'datacenter_id': int(dc_id),
                'allocated_mips': float(mips),
                'requested_ram': float(ram),
                'requested_bw': float(bw)
            })
        except:
            continue
    return vms_data

all_vms = []
for index, row in df.iterrows():
    vms = parse_vms(row['vms'])
    for vm in vms:
        vm['time'] = row['time']
        vm['host_id'] = row['host_id']
        all_vms.append(vm)

vms_df = pd.DataFrame(all_vms)
vm_summary_per_host = vms_df.groupby('host_id')[['allocated_mips', 'requested_ram', 'requested_bw']].sum().reset_index()

# === STEP 7: Datacenter-wide Totals & Usage ===
total_mips = df['mips'].sum()
available_mips = df['available_mips'].sum()
mips_usage_pct_dc = (total_mips - available_mips) / total_mips * 100 if total_mips else 0

total_ram = df['ram'].sum()
available_ram = df['available_ram'].sum()
ram_usage_pct_dc = (total_ram - available_ram) / total_ram * 100 if total_ram else 0

total_bw = df['bw'].sum()
available_bw = df['available_bw'].sum()
bw_usage_pct_dc = (total_bw - available_bw) / total_bw * 100 if total_bw else 0

total_storage = df['storage'].sum()
available_storage = df['available_storage'].sum()
storage_usage_pct_dc = (total_storage - available_storage) / total_storage * 100 if total_storage else 0

# === STEP 8: Interpolated CPU and RAM Energy for All Hosts ===
df['time_hours'] = df['time'] / 3600
total_cpu_energy_kwh = 0.0
total_ram_energy_kwh = 0.0

for host_id, group in df.groupby('host_id'):
    group_sorted = group.sort_values(by='time_hours')
    times = group_sorted['time_hours'].values
    cpu_powers = group_sorted['cpu_power'].values
    ram_powers = group_sorted['ram_power'].values

    for i in range(1, len(times)):
        delta_t = times[i] - times[i - 1]
        # CPU Energy
        avg_cpu_power = 0.5 * (cpu_powers[i] + cpu_powers[i - 1])
        total_cpu_energy_kwh += (avg_cpu_power * delta_t) / 1000
        # RAM Energy
        avg_ram_power = 0.5 * (ram_powers[i] + ram_powers[i - 1])
        total_ram_energy_kwh += (avg_ram_power * delta_t) / 1000

# === STEP 9: Summary Counts ===
num_active_hosts = active_hosts_df['host_id'].nunique()
total_vms = len(vms_df)
vms_per_host = vms_df.groupby('host_id').size().reset_index(name='vm_count')

# === DISPLAY RESULTS ===
print("=== Host-Level Resource Usage Summary ===")
print(f"Total active hosts (running): {num_active_hosts}")
print(f"Average CPU usage per host (%): {sum(cpu_usage_pct) / len(cpu_usage_pct):.2f}")
print(f"Average RAM usage per host (%): {sum(ram_usage_pct) / len(ram_usage_pct):.2f}")
print(f"Average Bandwidth usage per host (%): {sum(bw_usage_pct) / len(bw_usage_pct):.2f}")
print(f"Average Storage usage per host (%): {sum(storage_usage_pct) / len(storage_usage_pct):.2f}")

print("\n=== Power Analysis ===")
print(f"Total CPU Power (W): {total_cpu_power:.2f}")
print(f"Total RAM Power (W): {total_ram_power:.2f}")

print("\n=== Datacenter-Wide Totals and Usage ===")
print(f"Total MIPS: {total_mips:.2f}, Usage: {mips_usage_pct_dc:.2f}%")
print(f"Total RAM: {total_ram:.2f}, Usage: {ram_usage_pct_dc:.2f}%")
print(f"Total Bandwidth: {total_bw:.2f}, Usage: {bw_usage_pct_dc:.2f}%")
print(f"Total Storage: {total_storage:.2f}, Usage: {storage_usage_pct_dc:.2f}%")

print("\n=== VM Summary ===")
print(f"Total VMs parsed: {total_vms}")
print("VMs per host:")
print(vms_per_host)

print("\n=== Interpolated Energy Consumption ===")
print(f"Total CPU Energy (kWh): {total_cpu_energy_kwh:.6f}")
print(f"Total RAM Energy (kWh): {total_ram_energy_kwh:.6f}")
```

```python
# Make sure time is in hours
df['time_hours'] = df['time'] / 3600

idle_energy_kwh = 0.0
active_energy_kwh = 0.0

# Group by host and integrate each separately
for host_id, group in df.groupby('host_id'):
    group = group.sort_values(by='time_hours')
    times = group['time_hours'].values
    powers = group['cpu_power'].values
    activity = group['active'].values  # True/False

    for i in range(1, len(times)):
        delta_t = times[i] - times[i - 1]
        avg_power = 0.5 * (powers[i] + powers[i - 1])

        # Check if both samples are active or idle
        if activity[i] and activity[i - 1]:
            active_energy_kwh += (avg_power * delta_t) / 1000
        elif not activity[i] and not activity[i - 1]:
            idle_energy_kwh += (avg_power * delta_t) / 1000
        else:
            # One is active, the other idle — split contribution
            active_energy_kwh += (avg_power * delta_t * 0.5) / 1000
            idle_energy_kwh += (avg_power * delta_t * 0.5) / 1000

# Output results
print(f"Interpolated ACTIVE CPU energy: {active_energy_kwh:.6f} kWh")
print(f"Interpolated IDLE  CPU energy: {idle_energy_kwh:.6f} kWh")
```

```python
# --- Step 1: Extract idle and max power per host from power_model column
def extract_idle_max(power_model_str):
    if pd.isna(power_model_str):
        return (np.nan, np.nan)
    try:
        power_vals = [float(x) for x in power_model_str.strip(',').split(',')]
        return (power_vals[0], power_vals[-1])  # idle, max
    except:
        return (np.nan, np.nan)

df[['p_idle', 'p_max']] = df['power_model'].apply(lambda s: pd.Series(extract_idle_max(s)))

# --- Step 2: Calculate MIPS-based utilization
df['mips_utilization'] = (df['mips'] - df['available_mips']) / df['mips']

# --- Step 3: Estimate CPU power from utilization and power model
df['estimated_cpu_power'] = df['p_idle'] + (df['p_max'] - df['p_idle']) * df['mips_utilization']

# --- Step 4: Convert time to hours
df['time_hours'] = df['time'] / 3600

# --- Step 5: Interpolate energy per host, grouped by datacenter
datacenter_energy = {}

for (dc_id, host_id), group in df.groupby(['datacenter_id', 'host_id']):
    group_sorted = group.sort_values(by='time_hours')
    times = group_sorted['time_hours'].values
    powers = group_sorted['estimated_cpu_power'].values

    host_energy_kwh = 0.0
    for i in range(1, len(times)):
        delta_t = times[i] - times[i - 1]
        avg_power = 0.5 * (powers[i] + powers[i - 1])
        host_energy_kwh += (avg_power * delta_t) / 1000

    # Add to datacenter total
    if dc_id not in datacenter_energy:
        datacenter_energy[dc_id] = 0.0
    datacenter_energy[dc_id] += host_energy_kwh

# --- Step 6: Print per-datacenter totals and overall total
print("=== Estimated CPU Energy by Datacenter (kWh) ===")
total_energy_kwh = 0.0
for dc_id, energy in datacenter_energy.items():
    print(f"Datacenter {dc_id}: {energy:.6f} kWh")
    total_energy_kwh += energy

print(f"\nTotal Estimated CPU Energy (All Datacenters): {total_energy_kwh:.6f} kWh")
```
